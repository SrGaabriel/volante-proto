```kotlin
%use dataframe

import org.jetbrains.kotlinx.dataframe.DataFrame
import org.jetbrains.kotlinx.dataframe.io.readExcel

val dataframe = DataFrame.readExcel("result.xlsx")
```

```kotlin
%use kandy
```

```kotlin
val xCo = "Esperado - xG"
val yCo = "Progressão - PrgC"

val converted = dataframe.convert(*arrayOf(xCo,yCo)).with { it.toString().toDoubleOrNull() }

converted.plot { 
    points {
        x(xCo)
        y(yCo)
        // Make it so when a point is hovered, the player name shows
        tooltips("Jogador")
    }
}
```
